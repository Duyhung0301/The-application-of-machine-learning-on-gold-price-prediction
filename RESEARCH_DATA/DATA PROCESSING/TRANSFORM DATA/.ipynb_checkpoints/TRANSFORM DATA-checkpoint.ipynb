# CPI VIETNAM

```python
import pandas as pd

# Đọc dữ liệu từ file Excel
file_path = "cpi_vietnam.xlsx"  # Đảm bảo file này có sẵn trong thư mục chạy
cpi_df = pd.read_excel(file_path, index_col=0)

# Tạo danh sách ngày từ 1/1/2010 đến 31/12/2024
date_range = pd.date_range(start="2010-01-01", end="2024-12-31", freq='D')

# Chuyển đổi dữ liệu CPI từ bảng tháng sang dạng dài
cpi_long = cpi_df.reset_index().melt(id_vars=["Tháng trước = 100"], var_name="Year", value_name="CPI")

# Định nghĩa ánh xạ tên tháng thành số
month_mapping = {"Tháng 1": 1, "Tháng 2": 2, "Tháng 3": 3, "Tháng 4": 4, "Tháng 5": 5, "Tháng 6": 6,
                 "Tháng 7": 7, "Tháng 8": 8, "Tháng 9": 9, "Tháng 10": 10, "Tháng 11": 11, "Tháng 12": 12}
cpi_long["Month"] = cpi_long["Tháng trước = 100"].map(month_mapping)

# Chuyển đổi thành ngày đầu tháng
cpi_long["Date"] = pd.to_datetime(cpi_long[["Year", "Month"]].assign(day=1))

# Tạo DataFrame chứa danh sách ngày
date_df = pd.DataFrame({"Date": date_range})

# Gộp dữ liệu để gán CPI cho từng ngày
merged_df = date_df.merge(cpi_long[["Date", "CPI"]], on="Date", how="left")

# Điền CPI cho tất cả các ngày trong tháng bằng forward fill
merged_df["CPI"] = merged_df["CPI"].fillna(method="ffill")

# Xuất file Excel
output_file = "CPI_Daily_Vietnam.xlsx"
merged_df.to_excel(output_file, index=False)

print(f"File {output_file} đã được tạo thành công!")
```

# VÀNG THẾ GIỚI

```python
import pandas as pd

# Đọc dữ liệu từ file Excel
file_path = "world_gold_price.xlsx"  # Đảm bảo file này có sẵn trong thư mục chạy
gold_df = pd.read_excel(file_path)

# Chuyển đổi cột "Time" sang định dạng datetime
gold_df["Time"] = pd.to_datetime(gold_df["Time"], format="%d/%m/%Y")

# Tạo danh sách ngày liên tục từ 1/1/2010 đến 31/12/2024
date_range = pd.date_range(start="2010-01-01", end="2024-12-31", freq='D')
date_df = pd.DataFrame({"Time": date_range})

# Gộp dữ liệu để có danh sách ngày liên tục
gold_full_df = date_df.merge(gold_df, on="Time", how="left")

# Xuất file Excel
output_file = "new.xlsx"
gold_full_df.to_excel(output_file, index=False)

print(f"File {output_file} đã được tạo thành công!")
```
