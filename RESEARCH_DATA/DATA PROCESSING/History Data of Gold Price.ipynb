```python
# Đường dẫn đến ChromeDriver
service = Service('chromedriver-win64/chromedriver.exe')  # Đường dẫn đúng đến chromedriver.exe
driver = webdriver.Chrome(service=service)

try:
    # Mở URL
    driver.get("https://giavang.org/trong-nuoc/sjc/lich-su/")
    time.sleep(random.randint(5, 10))  # Đợi trang tải

    # Chờ cho đến khi các phần tử được hiển thị
    WebDriverWait(driver, 10).until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, ".col-xs-12.col-sm-6.col-md-4.col-lg-3 a")))

    # Tìm tất cả các phần tử với class chỉ định
    elems = driver.find_elements(By.CSS_SELECTOR, ".col-xs-12.col-sm-6.col-md-4.col-lg-3 a")

    # Lấy tiêu đề và liên kết từ các phần tử
    title = [elem.get_attribute('title') for elem in elems]
    links = [elem.get_attribute('href') for elem in elems]

    # Gán thông tin vào DataFrame
    df1 = pd.DataFrame(list(zip(title, links)), columns=['title', 'link_item'])
    df1['index_'] = np.arange(1, len(df1) + 1)

    # Lưu df1 vào file CSV
    df1.to_csv('df1.csv', index=False)
    print("Dữ liệu đã được lưu vào file 'df1.csv'")
    print(df1)

finally:
    driver.quit()  # Đảm bảo trình duyệt được đóng sau khi hoàn thành
```

```python
# Danh sách lưu giá mua vào và bán ra
mua_vao = []
ban_ra = []
# Đường dẫn đến ChromeDriver
service = Service('chromedriver-win64/chromedriver.exe')  # Đường dẫn đúng đến chromedriver.exe
driver = webdriver.Chrome(service=service)

# Truy cập vào 5 link đầu tiên trong df1
for index, row in df1.iterrows():
    if index >= 5:  # Chỉ lấy 5 dòng đầu tiên
        break
    links = row['link_item']
    driver.get(links)
    sleep(2)  # Đợi trang load

    # Kiểm tra nếu có thẻ <div class="box-content">
    try:
        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.XPATH, '/html/body/main/div/section[1]')))
        
        # Kiểm tra sự tồn tại của thẻ <div class="box-content">
        if len(driver.find_elements(By.XPATH, '/html/body/main/div/section[1]/div[@class="box-content"]')) > 0:
            # Nếu có thẻ <div class="box-content">
            mua_vao.append(None)
            ban_ra.append(None)
        else:
            # Nếu không có thẻ <div class="box-content">
            # Lấy giá mua vào
            mua_vao_value = driver.find_element(By.XPATH, '/html/body/main/div/section[1]/div/div[1]/div[1]/div/span[2]').text
            mua_vao.append(mua_vao_value)
            
            # Lấy giá bán ra
            ban_ra_value = driver.find_element(By.XPATH, '/html/body/main/div/section[1]/div/div[1]/div[2]/div/span[2]').text
            ban_ra.append(ban_ra_value)
    
    except (NoSuchElementException, ElementNotInteractableException) as e:
        print(f"Lỗi truy xuất dữ liệu: {e}")
        mua_vao.append(None)
        ban_ra.append(None)
    
    time.sleep(random.uniform(1, 3))  # Ngủ một chút trước khi truy cập link tiếp theo

# Đóng WebDriver
driver.quit()

# Chuyển đổi danh sách thành DataFrame để dễ xử lý
df2 = pd.DataFrame({'mua_vao': mua_vao, 'ban_ra': ban_ra})
print(df2)
```

```python
import numpy as np
from selenium import webdriver
import random
from selenium.common.exceptions import NoSuchElementException, ElementNotInteractableException
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
import pandas as pd
import time
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Tự động tải ChromeDriver phù hợp với phiên bản Chrome
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

df1 = pd.read_csv('df1.csv')

try:
    saved_data = pd.read_csv('gia_vang_sjc.csv')
    saved_indices = saved_data.index.tolist()  
except FileNotFoundError:
    saved_indices = []  

mua_vao = []
ban_ra = []

for index, row in df1.iterrows():
    if index < 5577:  
        continue

    link = row['link_item']
    title = row['title']  
    driver.get(link)
    
    time.sleep(random.uniform(5, 10))

    try:
        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.XPATH, '/html/body/main/div/section[1]')))
        
        if len(driver.find_elements(By.XPATH, '/html/body/main/div/section[1]/div[@class="box-content"]')) > 0:
            mua_vao.append(None)
            ban_ra.append(None)
        else:
            mua_vao_value = driver.find_element(By.XPATH, '/html/body/main/div/section[1]/div/div[1]/div[1]/div/span[2]').text
            mua_vao.append(mua_vao_value)

            ban_ra_value = driver.find_element(By.XPATH, '/html/body/main/div/section[1]/div/div[1]/div[2]/div/span[2]').text
            ban_ra.append(ban_ra_value)

    except (NoSuchElementException, ElementNotInteractableException) as e:
        print(f"Lỗi truy xuất dữ liệu tại index {index}: {e}")
        mua_vao.append(None)
        ban_ra.append(None)
    
    except Exception as e:
        print(f"Lỗi khác tại index {index}: {e}")
        mua_vao.append(None)
        ban_ra.append(None)

    df_temp = pd.DataFrame({'ngay': [title], 'mua_vao': [mua_vao[-1]], 'ban_ra': [ban_ra[-1]]})

    df_temp.to_csv('gia_vang_sjc.csv', index=False, mode='a', header=False) 

    time.sleep(random.uniform(5, 10))

driver.quit()

print("oke xong ròi đóa hihi")
```
