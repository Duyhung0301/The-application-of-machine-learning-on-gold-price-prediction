```python
import pandas as pd

file_path = 'raw_data_nckh.xlsx'
data = pd.read_excel(file_path)

print("Dữ liệu ban đầu:")
print(data.head())

data_filled = data.fillna(method='ffill')

# Kiểm tra dữ liệu sau khi xử lý
print("\nDữ liệu sau khi điền giá trị thiếu:")
print(data_filled.head())

# Lưu dữ liệu đã xử lý vào file Excel mới nếu cần
data_filled.to_excel('nckh_data_processed.xlsx', index=False)

print("\nDữ liệu đã được lưu vào file 'nckh_data_processed.xlsx'.")
```
